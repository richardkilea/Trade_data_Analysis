```python
import pandas as pd
import numpy as np
import glob
```

```python
columns_to_load = ["period","reporterISO", "reporterDesc", "flowCode","partnerDesc","partnerISO", "cmdCode", "H5","cifvalue", "fobvalue"]
```

```python
root_folder = "/content/drive/MyDrive/Auto Research Team Folder/UNCOMTRADE Data/HS 6 DIGIT/2022/Converted To 2017 H5"
file_pattern = root_folder + "/*.csv"
files = glob.glob(file_pattern)
# Check matched files
print(files)
```

```python
dataframes = []
i=0
```

```python
for file in files:
    i+=1
    print(f"Loading {file}...")
    # Read the CSV file with specified columns
    df_chunk = pd.read_csv(file, usecols=columns_to_load)
    # Append the resulting DataFrame to the list
    dataframes.append(df_chunk)
```

```python
final_df_2022 = pd.concat(dataframes, ignore_index=True)
```

```python
final_df_2022.shape[0]
```

```python
final_df_2022.head()
```

```python
print(final_df_2022['cifvalue'].sum())
print(final_df_2022['fobvalue'].sum())
```

```python
# Replace non-finite values with a suitable integer (e.g., -1) before conversion
final_df_2022['H5'] = pd.to_numeric(final_df_2022['H5'], errors='coerce').fillna(-1).astype(int)
```

```python
final_df_2022['H5'] = final_df_2022['H5'].astype(str).str.zfill(6)  # Pad with leading zeros
final_df_2022['cmdCode'] = final_df_2022['cmdCode'].astype(str).str.zfill(6)  # Pad with leading zeros
```

```python
final_df_2022.sort_values(by='cmdCode', ascending=True, inplace=True)
```

```python
final_df_2022.head()
```

```python
flow_codes = final_df_2022['flowCode'].unique()
split_dfs = {}
# Iterate through unique flow codes and split the DataFrame
for flow_code in flow_codes:
  split_dfs[flow_code] = final_df_2022[final_df_2022['flowCode'] == flow_code]
```

split dataframes based of flowcode

```python
import_df = split_dfs['M']
export_df = split_dfs['X']
```

```python
import_df = import_df[import_df.partnerDesc != 'World']
export_df = export_df[export_df.partnerDesc != 'World']
```

```python
import_df.reset_index(drop=True, inplace=True)
export_df.reset_index(drop=True, inplace=True)
```

```python
import_df = import_df.drop(columns=['fobvalue', 'flowCode'])
export_df = export_df.drop(columns=['cifvalue', 'flowCode'])
```

```python
export_df = export_df.rename(columns={'reporterISO':'exporterISO','reporterDesc':'exporter','partnerISO':'importerISO','partnerDesc':'importer','fobvalue': 'export_value'})
import_df = import_df.rename(columns={'reporterISO': 'importerISO','reporterDesc':'importer','partnerISO':'exporterISO','partnerDesc':'exporter','cifvalue': 'import_value'})
```
