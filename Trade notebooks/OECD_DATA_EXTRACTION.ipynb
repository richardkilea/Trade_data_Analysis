parse the XML response content and define the namespaces

```python
# Define the namespaces for the XML
namespaces = {
    'generic': 'http://www.sdmx.org/resources/sdmxml/schemas/v2_1/data/generic',
    'message': 'http://www.sdmx.org/resources/sdmxml/schemas/v2_1/message'
}
```

```python
# Parse the XML content
tree = ET.ElementTree(ET.fromstring(response.content))
root = tree.getroot()
```

initialize the list where we’ll collect data for the DataFrame

```python
# Prepare a list to collect data for the DataFrame
data_points = []
```

iterate over each series in the XML, extracts the dimensions from SeriesKey, and captures each observation (Obs).

```python
# Iterate through each series in the dataset
for series in root.findall('.//generic:Series', namespaces):
    # Initialize a dictionary to hold dimension data for each series
    series_data = {}

  # Extract all dimensions from SeriesKey and add them to series_data
    for value in series.find('generic:SeriesKey', namespaces):
        dimension_id = value.attrib.get('id')
        dimension_value = value.attrib.get('value')
        series_data[dimension_id] = dimension_value

    # Iterate over observations in the series
    for obs in series.findall('.//generic:Obs', namespaces):
        time_period_elem = obs.find('generic:ObsDimension', namespaces)
        value_elem = obs.find('generic:ObsValue', namespaces)

        # Check that elements exist before accessing their values
        if time_period_elem is not None and value_elem is not None:
            # Add time period and observation value to series_data
            series_data['TimePeriod'] = time_period_elem.attrib.get('value')
            series_data['Value'] = value_elem.attrib.get('value')

            # Append a copy of series_data for this observation to the data list
            data_points.append(series_data.copy())
```

```python
# Convert the list of dictionaries to a DataFrame
df = pd.DataFrame(data_points)
df.head()
```
